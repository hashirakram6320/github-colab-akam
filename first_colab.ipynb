```python
import pandas as pd
data = pd.read_csv("robot_mission.csv")
numbers = data["number"].tolist()
rounds = data["round"].tolist()
print(numbers)
print(rounds)
```

```python
total = 0
positions = []

for i in range(len(numbers)):
    current = numbers[i]

    if current % 2 == 0:
        total += current
    else:
        total -= current

    positions.append(total)

    print("Current number:", current)
    print("Robot's new position:", total)

import matplotlib.pyplot as plt

plt.plot(rounds, positions, marker='.')
plt.axhline(color='r', linestyle='-', linewidth=1)
plt.xlabel("Rounds")
plt.ylabel("Positions")
plt.title("Robot's Journey")
plt.xticks(rounds)
plt.grid(True)
plt.show()

print("Robot's final position:", total)
```
